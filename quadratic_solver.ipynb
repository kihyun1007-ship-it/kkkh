# 이차방정식 풀이 노트북

다음 셀들을 실행해 원하는 계수로 이차방정식을 풀어보세요.

```python
import cmathdef solve_quadratic(a: float, b: float, c: float):    """계수 a, b, c를 받아 이차방정식 ax^2 + bx + c = 0의 해를 반환합니다.    복소 근도 지원합니다.    """    if a == 0:        raise ValueError('a는 0이 될 수 없습니다.')    discriminant = b ** 2 - 4 * a * c    sqrt_disc = cmath.sqrt(discriminant)    x1 = (-b + sqrt_disc) / (2 * a)    x2 = (-b - sqrt_disc) / (2 * a)    return discriminant, (x1, x2)print('함수를 불러왔습니다. `solve_quadratic(a, b, c)`를 호출해보세요.')
```

## 사용 예시
`a = 1, b = -3, c = 2`인 경우를 풀어봅니다.

```python
disc, roots = solve_quadratic(1, -3, 2)print(f'판별식: {disc}')print(f'해: {roots[0]}, {roots[1]}')
```

## 직접 계수를 넣어보세요
